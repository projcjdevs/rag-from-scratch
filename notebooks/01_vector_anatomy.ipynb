**NumPy Implementation**

```python
import numpy as np

# A simple 5-dimensional vector 
v = np.array([3.0, 2.6, -1.4, 0.8, -2.1])

print("Vector v:", v)
print("Shape:", v.shape)
print("Number of dimensions:", v.ndim)
```

**Implement L2 Norm from scratch**

    Computes the Euclidean (L2) magnitude of a vector.
    Formula: ||v|| = sqrt( sum(v_i^2) )
    This is the Pythagorean theorem generalized to n dimensions.

```python
def l2_norm_scratch(v: np.ndarray) -> float: 

    sum_of_squares = np.sum(v ** 2)   # v_1^2 + v_2^2 + ... + v_n^2
    magnitude = np.sqrt(sum_of_squares)
    return magnitude

magnitude = l2_norm_scratch(v)
print(f"Manual L2 Norm: {magnitude}")

print(f"NumPy L2 Norm: {np.linalg.norm(v):.14f}") # We will use NumPy's built-in function to compare with scratch function
print(f"Match: {np.isclose(magnitude, np.linalg.norm(v))}")
```

**Implement Normalization**

    Returns the unit vector (normalized version) of v.
    
    Formula: v_hat = v / ||v||
    
    The result always has magnitude exactly 1.0 — it lands on the unit hypersphere.
    Dividing by magnitude is safe because embedding vectors are never zero vectors.

```python
def normalize(v: np.ndarray) -> np.ndarray:

    magnitude = l2_norm_scratch(v)
    return v / magnitude

v_normalized = normalize(v)

print("Original Vector: ", v)
print("Normalized Vector: ", np.round(v_normalized, 4))
print()
print(f"Original Magnitude: {l2_norm_scratch(v):.6f}")
print(f"Normalized Magnitude: {l2_norm_scratch(v_normalized):.6f}") # Always 1.0
```

**Why Normalization Matters**

This cell demonstrates why *normalization* matters for search. We'll show that without it, a bigger vector dominates unfairly.

```python
sentence_a_embedding = np.array([0.6, 0.8, 0.0])
sentence_b_embedding = np.array([6.0, 8.0, 0.0])
query_embedding = np.array([0.58, 0.81, 0.05])

# Without Normalization
raw_score_a = np.dot(query_embedding, sentence_a_embedding)
raw_score_b = np.dot(query_embedding, sentence_b_embedding)

print("===WITHOUT NORMALIZATION===")
print(f"Score A: {raw_score_a:.4f}")
print(f"Score B: {raw_score_b:.4f}")

# With Normalization
norm_score_a = np.dot(normalize(query_embedding), normalize(sentence_a_embedding))
norm_score_b = np.dot(normalize(query_embedding), normalize(sentence_b_embedding))

print("===WITH NORMALIZATION===")
print(f"Score A: {norm_score_a:.4f}")
print(f"Score B: {norm_score_b:.4f}")
```
