```python
# Cell 1 - Setup and Load Documents
import sys
sys.path.append("../src")

from rag import LightweightRAG

# Load documents
with open("../data/documents.txt", "r") as f:
    documents = [line.strip() for line in f if line.strip()]

print(f"Loaded {len(documents)} documents")
```

```python
# Cell 2 — Index
rag = LightweightRAG(n_clusters=3)
rag.index(documents, adj_threshold=0.72)
```

```python
# Cell 3 — Vector search
query = "how do computers process information?"

print(f'Query: "{query}"\n')
print("=== Vector Search (IVF) ===")
results = rag.search(query, top_k=3)
for rank, (doc, score) in enumerate(results, 1):
    print(f"  {rank}. [{score:.4f}] {doc}")
```

```python
# Cell 4 — Graph search (hybrid)
print(f'Query: "{query}"\n')
print("=== Hybrid Graph Search ===")
results = rag.graph_search(query, top_k=3, hops=2)
for rank, (doc, score) in enumerate(results, 1):
    print(f"  {rank}. [{score:.4f}] {doc}")
```

```python
# Cell 5 — Visualise
result_indices = [
    documents.index(doc)
    for doc, _ in rag.search(query, top_k=3, use_ivf=False)
]

rag.visualise(query=query, highlight_indices=result_indices)
```
